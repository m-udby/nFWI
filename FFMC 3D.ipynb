```python
#FFMC CALCULATOR

def oldFF(F1):
    MtpDenom = 59.5 + F1
    try:
        Mtp = 147.2 * ((101 - F1)/MtpDenom)
    except ZeroDivisionError :
        Mtp = 0
    if Mtp < 0 :
        Mtp = 0
    else :
        return Mtp

def FFcalc(F1, T, RA, H, W):
    H = np.clip(H,0,100) #DRY PHASE CANNOT PROCESS NEGATIVES OR SUPERSAT CONDITIONS
    Mtp = oldFF(F1) #in the manual as F0
    #RAINFALL PHASE
    #precip adjust
    if RA > 0.5 :
        Pf = RA - 0.5
    else : Pf = RA
    #prev day precip balancing
    if Mtp > 0 and Mtp <= 150 :
        with np.errstate(divide='raise'):
            try:
                Mrt = Mtp + 42.5 * Pf * (e**(-100/(251-Mtp))) * (1 - (e**(-6.93/Pf)))
            except FloatingPointError:
                Mrt = 50 #NEUTRAL VALUE IN CASE OF FAILURE FOR NOW
    else :
        with np.errstate(divide='raise'):
            try:
                Mrt = Mtp + 42.5 * Pf * (e**(-100/(251-Mtp))) * (1 - (e**(-6.93/Pf))) + 0.0015 * ((Mtp - 150)**2) * (Pf**0.5)
            except FloatingPointError:
                Mrt = 50 #NEUTRAL VALUE IN CASE OF FAILURE FOR NOW
    if Mrt >= 250 :
        Mrt = 250
    else : pass
    #DRYING PHASE
    '''    
    if H >= 0 and H < 100 :
        He = (H**0.679)
    elif H > 100 :
        He = 100 :
    else :
        He = 0''' 
    
    Ed = 0.942 * (H**0.679) + (11 * (e**((H-100)/10)))+0.18*(21.1-T)*(1-e**-(0.115*H)) #ADDED NP ABS TO EXPONENTS TO TRY AND GROUND SCALARS
    #LOG DRY CHECK
    if Mrt > Ed :
        Ko = 0.424 *(1-(H/100)**1.7)+0.0694*(W**0.5)*(1-((100-H)/100)**8)
        Kd = (Ko * 0.581) * (e**(0.0365*T))
        m = Ed + (Mrt - Ed) * (10**(-Kd))
    else : 
        #In the event of Ed larger than previous moisture value, we calc wet phase content
        Ew = 0.618 * (H**0.753)+10*(e**((H-100)/10))+0.18*(21.1-T)*(1-e**(-0.115*H))
        try:
            Ko = 0.424 * (1-((100-H)/100)**1.7)+0.0694 * (W**0.5) * (1-((100-H)/100)**8)
        except OverflowError :
            Ko = (W**0.5) #WE NEED TO READ UP ON THIS EQUATION FURTHER
        Kd = Ko * 0.581 * (e**(0.0365*T))
        m = Ew - (Ew - Mrt) *(10**-Kd)# DOUBLE CHECK VARIABLES WITH ORIGINAL EQUATION SHEET
        if Ew <= Mrt or Ed >= Mrt :
            m = Mrt
        else : pass
        #We used Ko and Kd here again for readibility, but the manual has them as Ew and Kw
    FF = 59.5 * ((250-m)/(147.2+m))
    return Ed #FF

def FFloop(T, RA, H, W):
    startcheck = time.perf_counter()
    rff = xr.full_like(T, fill_value=np.nan)
    for lat_idx in range(T.shape[1]):
                for lon_idx in range(T.shape[2]):
                    F1 = 85
                    for time_idx in range(T.shape[0]):
                        T_val = T.values[time_idx, lat_idx, lon_idx]
                        RA_val = RA.values[time_idx, lat_idx, lon_idx]
                        H_val = H.values[time_idx, lat_idx, lon_idx]
                        W_val = W.values[time_idx, lat_idx, lon_idx]
                        runff = FFcalc(F1, T_val, RA_val, H_val, W_val)
                        rff.values[time_idx, lat_idx, lon_idx] = runff
                        F1 = runff
                checkpoint = timedelta(seconds=time.perf_counter() - startcheck)    
                print(f"Checkpoint {lat_idx} at {checkpoint}")
    finalfmc = rff.rename('ffmc')
    finalfmc.attrs["units"] = 'fine fuel'
    return finalfmc

print(f"Running the {now} FFM Code for {model}'s {mode} data")      
testrun = FFloop(T, RA, RH, W)
print(testrun)
try:
    testrun.to_netcdf(f'Nebraska_{now}_{mode}_{model}_FFMC.nc')
    print(f'Success! Saved as an .nc of the ({now}) {model} {mode}')
except Exception as ee:
    print(f'{ee}')

```
